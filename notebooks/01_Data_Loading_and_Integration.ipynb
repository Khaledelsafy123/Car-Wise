```python
import pandas as pd
from pathlib import Path
PROJECT_ROOT = Path(r"D:\INSTANT\ML_SV_PROJ")

RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"
```

```python
csv_files = {
    "audi": "audi.csv",
    "bmw": "bmw.csv",
    "ford": "ford.csv",
    "hyundai": "hyundi.csv",
    "mercedes": "merc.csv",
    "skoda": "skoda.csv",
    "toyota": "toyota.csv",
    "vauxhall": "vauxhall.csv",
    "volkswagen": "vw.csv"
}
```

Merge the datasets

```python
dataframes = []

for manufacturer, filename in csv_files.items():

    file_path = next(RAW_DATA_PATH.rglob(filename))

    temp_df = pd.read_csv(file_path)

    temp_df["manufacturer"] = manufacturer

    dataframes.append(temp_df)

df = pd.concat(dataframes, ignore_index=True)

print(f"Dataset Shape: {df.shape}")
```

```python
df = df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("Dataset shuffled successfully.")
```

save files in **vehicles.csv**

```python
output_path = RAW_DATA_PATH / "vehicles.csv"

df.to_csv(
    output_path,
    index=False
)

print(f"Dataset saved successfully at: {output_path}")
```
