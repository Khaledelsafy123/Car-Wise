```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(r"D:\INSTANT\ML_SV_PROJ")
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "vehicles.csv"

df = pd.read_csv(DATA_PATH)

print(f"Dataset Shape: {df.shape}")
```

```python
display(df.head(10))

display(df.sample(10, random_state=42))

df.info()
```

```python
display(df.describe(include="all").T)
```

```python
total_duplicates = df.duplicated().sum()

print(f"Total Duplicate Rows: {total_duplicates}")
print(f"Duplicate Percentage: {total_duplicates / len(df) * 100:.2f}%")
```

Remove Duplicates

```python
df = df.drop_duplicates(keep="first").reset_index(drop=True)

print(f"Removed duplicates: {total_duplicates}")
print(f"Remaining rows: {len(df)}")
```

# Tax Feature

```python
print("Rows with tax only:",
      (df["tax"].notna() & df["tax(£)"].isna()).sum())

print("Rows with tax(£) only:",
      (df["tax"].isna() & df["tax(£)"].notna()).sum())

print("Rows with both:",
      (df["tax"].notna() & df["tax(£)"].notna()).sum())

print("Rows with neither:",
      (df["tax"].isna() & df["tax(£)"].isna()).sum())
```

```python
df["tax"] = df["tax"].combine_first(df["tax(£)"])

df.drop(columns=["tax(£)"], inplace=True)

print("Missing tax values:", df["tax"].isna().sum())
print("Total columns:", df.shape[1])
```

# Year Feature

```python
suspicious_years = df[df["year"] > 2026]

print("Records with year > 2026:", len(suspicious_years))

display(
    suspicious_years[
        ["manufacturer", "model", "year", "price"]
    ]
)
```

```python
invalid_year_count = (df["year"] > 2026).sum()

df = df[df["year"] <= 2026].copy()

print("Removed records:", invalid_year_count)
print("New dataset shape:", df.shape)
```

```python
old_vehicles = df[df["year"] < 1980]

print("Records before 1980:", len(old_vehicles))

display(
    old_vehicles[
        ["manufacturer", "model", "year", "price", "mileage"]
    ].sort_values("year")
)
```

```python
display(df.loc[
    df["year"] < 1980
].T)
```

```python
invalid_old_years = df["year"] < 1980

df = df.loc[~invalid_old_years].copy()

print("Removed records:", invalid_old_years.sum())
print("Dataset Shape:", df.shape)
```

# MPG Feature

```python
display(
    df.nlargest(15, "mpg")[
        ["manufacturer", "model", "year", "price", "fuelType", "mpg"]
    ]
)
```

```python
display(
    df.nsmallest(15, "mpg")[
        ["manufacturer", "model", "year", "price", "fuelType", "mpg"]
    ]
)
```

```python
df["mpg"].describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
```

```python
low_mpg = df[df["mpg"] < 5]

print("Records below 5 MPG:", len(low_mpg))

display(
    low_mpg.groupby(["manufacturer", "model", "fuelType", "mpg"])
    .size()
    .reset_index(name="count")
    .sort_values("mpg")
)
```

```python
invalid_mpg = df["mpg"] < 1

df = df.loc[~invalid_mpg].copy()

print("Removed records:", invalid_mpg.sum())
print("Dataset Shape:", df.shape)
```

# Engine_Size Feature

```python
zero_engine = df[df["engineSize"] == 0]

print("Records with zero engine size:", len(zero_engine))

display(
    zero_engine.groupby(["manufacturer", "model", "fuelType"])
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
)
```

```python
invalid_engine = (
    (df["engineSize"] == 0) &
    ~((df["manufacturer"] == "bmw") & (df["model"] == "i3"))
)

df.loc[invalid_engine, "engineSize"] = np.nan

print("Converted to NaN:", invalid_engine.sum())
print("Remaining zero engine sizes:", (df["engineSize"] == 0).sum())
print("Missing engine sizes:", df["engineSize"].isna().sum())
```

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": df.isna().mean() * 100
})

display(missing_summary.sort_values("missing_count", ascending=False))
```

# Price Feature

```python
print(df["price"].describe())

print("\nPrices <= 0:", (df["price"] <= 0).sum())

display(
    df.nsmallest(15, "price")[
        ["manufacturer", "model", "year", "price", "mileage", "fuelType"]
    ]
)
```

```python
display(
    df.nlargest(15, "price")[
        ["manufacturer", "model", "year", "price",
         "mileage", "fuelType", "engineSize"]
    ]
)
```

```python
print("Data types:")
print(df[["manufacturer", "model", "price"]].dtypes)

print("\nIndividual conditions:")
print("Mercedes:", (df["manufacturer"] == "mercedes").sum())
print("A Class:", (df["model"] == "A Class").sum())
print("Price > 50000:", (df["price"] > 50000).sum())

print("\nMatching records:")
display(
    df[
        (df["manufacturer"] == "mercedes") &
        (df["model"] == "A Class") &
        (df["price"] > 50000)
    ]
)
```

# Model Feature

```python
mercedes_models = (
    df.loc[df["manufacturer"] == "mercedes", "model"]
    .value_counts()
)

display(mercedes_models.head(30))

print("\nModels containing 'Class':")
display(
    mercedes_models[
        mercedes_models.index.str.contains("Class", case=False, na=False)
    ]
)
```

```python
a_class_values = df.loc[
    df["manufacturer"] == "mercedes", "model"
]

for value in a_class_values.unique():
    if "A Class" in value:
        print("Value:", repr(value))
        print("Length:", len(value))
        print("Exact match:", value == "A Class")
```

```python
df["model"] = df["model"].str.strip()

print("A Class records:", (df["model"] == "A Class").sum())
print("Mercedes A Class > £50,000:", (
    (df["manufacturer"] == "mercedes") &
    (df["model"] == "A Class") &
    (df["price"] > 50000)
).sum())
```

```python
display(
    df[
        (df["manufacturer"] == "mercedes") &
        (df["model"] == "A Class") &
        (df["price"] > 50000)
    ][
        ["year", "price", "transmission", "mileage", "fuelType", "mpg", "engineSize"]
    ].sort_values("price", ascending=False)
)
```

```python
display(
    df[
        (df["manufacturer"] == "bmw") &
        (df["model"] == "2 Series") &
        (df["price"] > 100000)
    ]
)
```

```python
print("Tax statistics:")
display(df["tax"].describe())

print("\nNegative tax:", (df["tax"] < 0).sum())
print("Zero tax:", (df["tax"] == 0).sum())
print("Missing tax:", df["tax"].isna().sum())

print("\nLowest 10 tax values:")
display(df.nsmallest(10, "tax"))

print("\nHighest 10 tax values:")
display(df.nlargest(10, "tax"))
```
